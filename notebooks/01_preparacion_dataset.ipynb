# PC2 - 01. Preparación y revisión del dataset
Este notebook revisa las 50 imágenes: 5 identidades × 10 imágenes.

```
from pathlib import Path
import pandas as pd
ROOT=Path('/content/PC2_Sistema_Inteligente_Facial_FINAL')
META=pd.read_csv(ROOT/'dataset/metadata.csv')
META.head(), META.groupby('id_persona').size(), META['emocion_sugerida'].value_counts()
```

```
import matplotlib.pyplot as plt
from PIL import Image
fig,axs=plt.subplots(5,10,figsize=(18,9))
for i,(_,r) in enumerate(META.iterrows()):
    ax=axs[i//10,i%10]; ax.imshow(Image.open(ROOT/'dataset'/r['ruta'])); ax.axis('off'); ax.set_title(r['archivo'],fontsize=6)
plt.tight_layout()
```

**Nota metodológica:** las imágenes de este paquete son variaciones sintéticas generadas a partir de cinco retratos base. Para cumplir literalmente la instrucción de “captura propia”, deben sustituirse por fotografías tomadas por el grupo si el docente lo exige.
